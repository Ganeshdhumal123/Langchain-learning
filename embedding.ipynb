```python
from langchain_huggingface import HuggingFaceEmbeddings
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Initialize free Hugging Face model
embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2",
    model_kwargs={"device": "cpu"},
    encode_kwargs={"normalize_embeddings": True}
)

print("Embedding model loaded successfully!")
```

```python
documents = [
    "Virat Kohli is an Indian cricketer known for aggressive batting and leadership.",
    "MS Dhoni is a former Indian captain famous for his calm demeanor and finishing skills.",
    "Sachin Tendulkar is known as the God of Cricket and holds many batting records.",
    "Rohit Sharma is famous for elegant batting and record-breaking performances.",
    "Jasprit Bumrah is an Indian fast bowler known for his unique bowling action and yorkers."
]
```

```python
print("Generating document embeddings...")

doc_embeddings = embedding_model.embed_documents(documents)

print("Document embeddings generated!")
print("Number of documents:", len(doc_embeddings))
print("Embedding dimensions:", len(doc_embeddings[0]))
```

```python
def semantic_search(query, top_k=3):

    # Convert query into embedding
    query_embedding = embedding_model.embed_query(query)

    # Calculate cosine similarity
    scores = cosine_similarity(
        [query_embedding],
        doc_embeddings
    )[0]

    # Get top matching document indices
    top_indices = np.argsort(scores)[::-1][:top_k]

    print("\nQuery:", query)
    print("\nMost Relevant Documents:\n")

    for rank, index in enumerate(top_indices, start=1):

        print(f"Rank: {rank}")
        print(f"Similarity Score: {scores[index]:.4f}")
        print(f"Document: {documents[index]}")
        print("-" * 60)
```

```python
semantic_search("Tell me about Virat Kohli", top_k=3)
```
